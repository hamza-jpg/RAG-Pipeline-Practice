```python
# Generating Mock Files

# Generating TXT file
with open("test_text.txt", "w", encoding = "utf-8") as f:
  f.write("This is the first line.\nThis is the second line.\nThis is the third line")

# Generating CSV file
import csv
with open("test_data.csv", "w", newline = "", encoding = "utf-8") as f:
  writer = csv.writer(f)
  writer.writerow(["Name", "Age", "City"])
  writer.writerow(["Alice", "25", "New York"])
  writer.writerow(["Bob", "30", "San Francisco"])
  writer.writerow(["Charlie", "35", "Los Angeles"])

# Generating PDF file
from reportlab.pdfgen import canvas
c = canvas.Canvas("test_doc.pdf")
c.drawString(100, 750, "This is the first page of the PDF.")
c.showPage()
c.drawString(100, 750, "This is the second page of the PDF.")
c.save()

print("Test files are succesfully generated!")
```

```python
# Text Loader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("test_text.txt", encoding = "utf-8")
documents = loader.load()

print(f"Total number of documents: {len(documents)}")
print(f"Metadata of the first document: {documents[0].metadata}")
print(f"Content of the first document:\n{documents[0].page_content}")
```

```python
# CSV Loader

from langchain_community.document_loaders import CSVLoader

loader = CSVLoader("test_data.csv")
documents = loader.load()

print(f"Total number of documents: {len(documents)}")
print(f"Metadata of the first document: {documents[0].metadata}")
print(f"Content of the first document:\n{documents[0].page_content}")

print(f"Total number of documents: {len(documents)}")
print(f"Metadata of the second document: {documents[1].metadata}")
print(f"Content of the second document:\n{documents[1].page_content}")

# Each row of the CSV file is a distinct document
```

```python
# PDF Loader

from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("test_doc.pdf")
documents = loader.load()

print(f"Total number of documents: {len(documents)}")

for i, doc in enumerate(documents):
  print(f"\n--- Page {i+1} ---")
  print(f"Metadata: {doc.metadata}")
  print(f"Content:\n{doc.page_content.strip()}")

# 'page' key starts from 0 watch out
```

```python
# WebBaseLoader

from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://www.python.org/")
docs = loader.load()
print("Total number of documents", len(docs))
print("First 200 characters\n", docs[0].page_content[:200])
```

```python
# Directory Loader

from langchain_community.document_loaders import DirectoryLoader, TextLoader

loader = DirectoryLoader(".", glob = "*.txt", loader_cls = TextLoader)
docs = loader.load()

print(f"Total number of .txt docs in directory: {len(docs)}")
for doc in docs:
  print(f"File: {doc.metadata['source']}")
```
